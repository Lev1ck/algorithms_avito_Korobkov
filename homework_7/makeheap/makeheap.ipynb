# Задача 1. Makeheap

## Ход решения

В minheap родитель не больше своих детей, поэтому в корне лежит минимум. В `makeheap_n_log_n` иду по списку слева направо и поднимаю каждый новый элемент, пока он меньше родителя. Каждый элемент может подняться на $O(\log n)$ уровней, поэтому в худшем случае построение занимает $O(n\log n)$.

В `makeheap` иду от последнего родителя к корню. Дочерние поддеревья уже являются кучами, поэтому меняю текущий элемент с меньшим ребёнком, пока текущий элемент больше него. Элемент может спуститься на $h$ уровней, только если у его начальной вершины есть потомок на такой глубине. Для вершины с индексом $i$ самый левый такой потомок имеет индекс $2^h(i+1)-1$. Этот индекс меньше $n$ ровно у $\lfloor n/2^h\rfloor$ вершин. Суммирую эти количества по всем уровням и получаю верхнюю границу для числа шагов спуска.

$$
\sum_{h=1}^{\infty}\left\lfloor\frac{n}{2^h}\right\rfloor
\le n\sum_{h=1}^{\infty}\frac{1}{2^h}=n.
$$

На начальные проверки родителей уходит ещё $O(n)$, поэтому `makeheap` работает за $O(n)$. Обе функции перестраивают исходный список и используют $O(1)$ дополнительной памяти.

## Реализация

```python
def _sift_down_min(arr, root, stop):
    """Восстанавливает minheap под root в пределах arr[0:stop]."""
    while True:
        left = 2 * root + 1
        if left >= stop:
            return

        right = left + 1
        child = left
        if right < stop and arr[right] < arr[left]:
            child = right

        if not arr[child] < arr[root]:
            return

        arr[root], arr[child] = arr[child], arr[root]
        root = child


def makeheap_n_log_n(arr):
    """Строит minheap подъёмами в O(n log n) и возвращает исходный список."""
    for index in range(1, len(arr)):
        child = index
        while child > 0:
            parent = (child - 1) // 2
            if not arr[child] < arr[parent]:
                break
            arr[parent], arr[child] = arr[child], arr[parent]
            child = parent
    return arr


def makeheap(arr):
    """Строит minheap просеиваниями вниз в O(n) и возвращает исходный список."""
    stop = len(arr)
    for root in range(stop // 2 - 1, -1, -1):
        _sift_down_min(arr, root, stop)
    return arr
```

## Пошаговый пример

Исходный список `[7, 2, 9, 4, 1, 8, 5, 3]`. В таблицах указаны все итерации внешнего цикла и обмены по индексам.

| Индекс вставки | Обмены при подъёме | Список после подъёма |
| ---: | --- | --- |
| 1 | 1 ↔ 0 | `[2, 7, 9, 4, 1, 8, 5, 3]` |
| 2 | — | `[2, 7, 9, 4, 1, 8, 5, 3]` |
| 3 | 3 ↔ 1 | `[2, 4, 9, 7, 1, 8, 5, 3]` |
| 4 | 4 ↔ 1, 1 ↔ 0 | `[1, 2, 9, 7, 4, 8, 5, 3]` |
| 5 | 5 ↔ 2 | `[1, 2, 8, 7, 4, 9, 5, 3]` |
| 6 | 6 ↔ 2 | `[1, 2, 5, 7, 4, 9, 8, 3]` |
| 7 | 7 ↔ 3 | `[1, 2, 5, 3, 4, 9, 8, 7]` |

| Корень просеивания | Обмены при спуске | Список после спуска |
| ---: | --- | --- |
| 3 | 3 ↔ 7 | `[7, 2, 9, 3, 1, 8, 5, 4]` |
| 2 | 2 ↔ 6 | `[7, 2, 5, 3, 1, 8, 9, 4]` |
| 1 | 1 ↔ 4 | `[7, 1, 5, 3, 2, 8, 9, 4]` |
| 0 | 0 ↔ 1, 1 ↔ 4 | `[1, 2, 5, 3, 7, 8, 9, 4]` |

У полученных куч разные расположения элементов. В обоих случаях каждый родитель меньше либо равен своим детям и в корне находится минимум `1`.

```python
example = [7, 2, 9, 4, 1, 8, 5, 3]
for build in (makeheap_n_log_n, makeheap):
    heap = example.copy()
    assert build(heap) is heap
    assert heap[0] == 1
    print(f"{build.__name__}(arr) → {heap}")
```

## Тесты

Для каждого результата проверяю все связи родитель–ребёнок, сохранение кратностей, минимум в корне непустой кучи и идентичность входного и возвращённого списков.

Использую типичные входы, граничные случаи и строки как пример других сравнимых значений. Перебираю все 1 093 списка длины от 0 до 6 из значений −1, 0 и 1. Дополнительно генерирую 2 000 списков с фиксированным зерном, повторами и отрицательными числами. Отдельно проверяю возрастающий, убывающий и полностью одинаковый списки длины 100 000.

```python
from itertools import product
from random import Random
import sys
import unittest


HEAP_BUILDERS = (makeheap_n_log_n, makeheap)


def is_minheap(items):
    return all(not items[i] < items[(i - 1) // 2] for i in range(1, len(items)))


EXAMPLES = (
    [],
    [42],
    [2, 1],
    [1, 2],
    list(range(15)),
    list(range(15, -1, -1)),
    [5] * 17,
    [0, -3, 2, -3, 0, 2, -1],
    [7, 2, 9, 4, 1, 8, 5, 3],
    [3, 2, 3, 1, 2, 4, 5, 5, 6, 4, -2],
    ["б", "в", "а", "а"],
)


def small_inputs():
    for size in range(7):
        for values in product((-1, 0, 1), repeat=size):
            yield list(values)


def random_inputs():
    rng = Random(314159)
    for _ in range(2_000):
        size = rng.randrange(129)
        yield [rng.randint(-30, 30) for _ in range(size)]


def large_inputs():
    size = 100_000
    return (
        ("ascending", list(range(size))),
        ("descending", list(range(size - 1, -1, -1))),
        ("equal", [5] * size),
    )


class MakeHeapTests(unittest.TestCase):
    def check_heap(self, values):
        for build in HEAP_BUILDERS:
            with self.subTest(method=build.__name__, n=len(values)):
                heap = values.copy()
                self.assertIs(build(heap), heap)
                self.assertTrue(is_minheap(heap))
                self.assertCountEqual(heap, values)
                if values:
                    self.assertEqual(heap[0], min(values))

    def test_examples(self):
        for values in EXAMPLES:
            with self.subTest(values=values):
                self.check_heap(values)

    def test_exhaustive_small_inputs(self):
        for values in small_inputs():
            self.check_heap(values)

    def test_seeded_random_inputs(self):
        for case, values in enumerate(random_inputs()):
            with self.subTest(case=case):
                self.check_heap(values)

    def test_large_patterns(self):
        for name, values in large_inputs():
            with self.subTest(pattern=name):
                self.check_heap(values)
```

```python
suite = unittest.defaultTestLoader.loadTestsFromTestCase(MakeHeapTests)
result = unittest.TextTestRunner(stream=sys.stdout, verbosity=2).run(suite)
assert result.wasSuccessful()
```

## Эксперимент

Сравниваю медианы семи запусков в миллисекундах. Для каждого метода есть один прогревочный запуск. На четырёх размерах проверяю случайные числа с повторами, возрастающую и убывающую последовательности, а также списки из равных значений. Случайные входы беру из диапазона `[-n, n]` с фиксированным зерном `20261126`.

На каждом запуске метод получает свежую копию одного и того же входа. Порядок методов сдвигается между раундами. `perf_counter` измеряет только вызов функции. Копирование входа, проверка результата и освобождение возвращённого объекта выполняются за пределами замера. Числа в выводах относятся к сохранённому запуску, а ячейка с экспериментом позволяет повторить замеры.

```python
from collections import Counter
import platform
from statistics import median
from time import perf_counter


def benchmark_inputs():
    rng = Random(20261126)
    for size in (1_000, 4_000, 16_000, 64_000):
        yield "Случайный", [rng.randint(-size, size) for _ in range(size)]
        yield "По возрастанию", list(range(size))
        yield "По убыванию", list(range(size - 1, -1, -1))
        yield "Все равны", [5] * size


def run_benchmarks(algorithms):
    repeats = 7
    rows = []

    for pattern, source in benchmark_inputs():
        expected = Counter(source)
        times = {method: [] for method in algorithms}

        for round_index in range(-1, repeats):
            offset = round_index % len(algorithms)
            order = algorithms[offset:] + algorithms[:offset]
            for method in order:
                arr = source.copy()
                start = perf_counter()
                result = method(arr)
                elapsed_ms = (perf_counter() - start) * 1_000

                assert is_minheap(result) and Counter(result) == expected
                del result
                if round_index >= 0:
                    times[method].append(elapsed_ms)

        rows.append((len(source), pattern, *(median(times[method]) for method in algorithms)))
    return rows


def print_benchmarks(algorithms, rows):
    print(f"{platform.python_implementation()} {platform.python_version()}, "
          f"{platform.system()} {platform.release()} ({platform.machine()}). Медианы, мс.")
    widths = [max(len(method.__name__), 10) for method in algorithms]
    names = " | ".join(f"{method.__name__:>{width}}" for method, width in zip(algorithms, widths))
    print(f"{'n':>6} | {'Вход':<15} | {names}")
    print("-" * (27 + sum(widths) + 3 * (len(widths) - 1)))
    for size, pattern, *times in rows:
        values = " | ".join(f"{value:>{width}.3f}" for value, width in zip(times, widths))
        print(f"{size:>6} | {pattern:<15} | {values}")
```

```python
algorithms = (makeheap_n_log_n, makeheap)
benchmark_rows = run_benchmarks(algorithms)
print_benchmarks(algorithms, benchmark_rows)
```

## Выводы

- На убывающем массиве каждый новый элемент доходит до корня при подъёме. Отношение времени двух методов выросло с 3,6 при 1 000 элементах до 6,2 при 64 000. На самом большом убывающем входе подъёмы заняли 199,913 мс, а построение снизу вверх — 32,121 мс.
- На случайном массиве из 64 000 элементов медианы составили 32,396 и 28,296 мс соответственно. Большинство подъёмов завершилось раньше корня, поэтому разница меньше, чем на убывающих данных.
- На возрастающих и одинаковых значениях подъём прекращается при первом сравнении с родителем. Здесь обе реализации делают линейное число проверок. Гарантия $O(n)$ для `makeheap` следует из доказательства через сумму высот и действует для любого порядка входа.
